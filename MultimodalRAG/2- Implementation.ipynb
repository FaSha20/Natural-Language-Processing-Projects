## Encoding Functions

```python
def load_image_from_url(url, timeout=10):
    final_url = url
    if "fa.wikipedia.org" in url:
        headers = {"User-Aent": "ImageDatasetScript/1.0 (email@gmail.com)"}
        response = requests.get(url, headers=headers, timeout=10)
        soup = BeautifulSoup(response.text, "html.parser")
        div = soup.find("div", {"id": "file"})
        if div:
            a_tag = div.find("a")
            if a_tag:
                href = a_tag.get("href")
                final_url = urljoin("https:", href) if href.startswith("//") else href
    headers = {"User-Agent": "ImageDatasetScript/1.0 (email@gmail.com)"}
    response = requests.get(final_url, headers=headers, timeout=10)
    response.raise_for_status()
    return Image.open(BytesIO(response.content)).convert("RGB")
```

```python
# ---- Text Encoder ----
def embed_text(text: str):
    text_embedding = BGE_m3_model.encode(text)['dense_vecs']
    return text_embedding

# ---- Image Encoder ----
def embed_image(url: str):
    try:
        image = load_image_from_url(url)
        image = preprocessor(images=image, return_tensors="pt")["pixel_values"][0]
        
        vision_encoder.eval()
        image_embedding = vision_encoder(image.unsqueeze(0)).pooler_output[0]
        return image_embedding, url
        
    except Exception as e:
        return None, None
```

## Build the DATABASE

```python
PERSIST_DIR = "/content/chroma_scientists_db"
os.makedirs(PERSIST_DIR, exist_ok=True)
client = chromadb.PersistentClient(path=PERSIST_DIR, settings=Settings(allow_reset=True))

text_db  = client.get_or_create_collection(name="scientists_text",  metadata={"hnsw:space": "cosine"})
image_db = client.get_or_create_collection(name="scientists_image", metadata={"hnsw:space": "cosine"})
```

```python
def upsert_pair(idx, name, text, image_url, t_emb, i_emb):
    meta = {
        "pair_id": idx,
        "name": name,
        "text": text,
        "image_url": image_url if image_url is not None else "",
    }

    if t_emb is not None:
        text_db.add(
            ids=[f"text-{idx}"],
            embeddings=[t_emb.tolist()],
            metadatas=[meta],
        )

    if i_emb is not None:
        image_db.add(
            ids=[f"img-{idx}"],
            embeddings=[i_emb.tolist()],
            metadatas=[meta],
        )
```

```python
for idx, person in enumerate(data):
    # name = person.get("name")
    text = person.get("text_chunk", "").replace("\u200c", "")
    name = ' '.join(extract_person_names(text))
    t_emb = embed_text(name)

    image_urls = list(person.get("image", {}).values())
    i_emb = None
    fixed_url = None

    for image_url in image_urls:
        if image_url:
            i_emb, url = embed_image(image_url)
            if i_emb is not None:
                fixed_url = url
                break

    if fixed_url and i_emb is not None:
        upsert_pair(idx, name, text, fixed_url, t_emb, i_emb)
```

## Retrieval Functions

```python
def search_by_text(query: str, k=5):
    q_emb = embed_text(query)
    res = text_db.query(query_embeddings=[q_emb.tolist()], n_results=k, include=["metadatas", "distances"])
    return res

def search_by_image(image_url: str, k=5):
    q_emb, url = embed_image(image_url)
    if q_emb is None:
        return None
    res = image_db.query(query_embeddings=[q_emb.tolist()], n_results=k, include=["metadatas", "distances"])
    return res, url
```

```python
import numpy as np

def _postprocess_query_result(res):
    out = []
    if not res:
        return out
    md = res.get("metadatas", [[]])[0]
    ds = res.get("distances", [[]])[0]
    for m, d in zip(md, ds):
        m = dict(m) if m else {}
        m["distance"] = float(d)
        out.append(m)
    return out

def retrieve_for_text_query(query: str, k_text=5, k_image=5):
    names = extract_person_names(query)
    print(names)
    if len(names) == 0:
        names.append(query)
    image_hits = []
    text_hits = []
    for query in names:
        q_emb = embed_text(query)
        if k_image>0:
            t_res = text_db.query(query_embeddings=[q_emb.tolist()], n_results=k_text, include=["metadatas", "distances"])
            i_res = image_db.query(query_embeddings=[q_emb.tolist()], n_results=k_image, include=["metadatas", "distances"])
            text_hits.append(_postprocess_query_result(t_res))
            image_hits += _postprocess_query_result(i_res)
        else:
            t_res = text_db.query(query_embeddings=[q_emb.tolist()], n_results=max(1, int(np.ceil(k_text//len(names)))), include=["metadatas", "distances"])
            text_hits += _postprocess_query_result(t_res)

    return text_hits, image_hits

def retrieve_for_image_query(image_url: str, k_text=5, k_image=5):
    q_emb, url = embed_image(image_url)
    if q_emb is None:
        return [], [], []
    elif k_text>0:
        t_res = text_db.query(query_embeddings=[q_emb.tolist()], n_results=k_text, include=["metadatas", "distances"])
        i_res = image_db.query(query_embeddings=[q_emb.tolist()], n_results=k_image, include=["metadatas", "distances"])
        text_hits  = _postprocess_query_result(t_res)
        image_hits = _postprocess_query_result(i_res)
        return text_hits, image_hits, url
    else:
        i_res = image_db.query(query_embeddings=[q_emb.tolist()], n_results=k_image, include=["metadatas", "distances"])
        
        image_hits = _postprocess_query_result(i_res)
        return [], image_hits, url
```

```python
from transformers import AutoModelForCausalLM

def gemma_rag_generate(
    question_text: str,
    options,
    text_hits,
    image_hits,
    url=None,
    max_new_tokens: int = 256,
    temperature: float = 0.9,
    top_p: float = 0.9,
):
    system_prompt = ("""
    با توجه به اطلاعات بازیابی شده به پرسش مطرح شده پاسخ بده و یکی از گزینه ها را به عنوان پاسخ انتخاب کن و دلیل خود را توضیح بده.
    درصورت نیاز از اطلاعات موجود در تصاویر هم استفاده کن"
    """
    )

    # Collect aligned pairs
    records = []
    if text_hits:
        for h in text_hits:
            records.append({
                "text": f"- {h.get('name','?')}: {h.get('text','')}",
                "image": h.get('image_url', None)
            })
    if image_hits:
        for h in image_hits:
            records.append({
                "text": f"- {h.get('name','?')}: {h.get('text','')}",
                "image": h.get('image_url', None)
            })

    # Add each record as a separate user message
    msg_content = []
    for rec in records:
        if rec["text"]:
            msg_content.append({"type": "text", "text": rec["text"]})
        if rec["image"]:
            img_obj = load_image_from_url(rec["image"])
            if img_obj:
                msg_content.append({"type": "image", "image": img_obj})
    if url:
        img = load_image_from_url(url)
        msg_content.append({"type": "text", "text": f"با توجه به عکس زیر، به این پرسش پاسخ بده: {question_text}\nگزینه‌ها:\n" + "\n".join(options)})
        msg_content.append({"type": "image", "image": img})
    else:
        msg_content.append({"type": "text", "text": f"پرسش: {question_text}\nگزینه‌ها:\n" + "\n".join(options)})
    
    messages = [{"role": "system",
                 "content": [{"type": "text", "text": system_prompt}]},
                {"role": "user", "content": msg_content}]

    # Apply chat template
    inputs = processor.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt"
    ).to(model.device, dtype=torch.bfloat16)

    input_len = inputs["input_ids"].shape[-1]
    with torch.inference_mode():
        out = generator_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=top_p
        )[0][input_len:]

    text = processor.decode(out, skip_special_tokens=True)

    return {
        "answer": text.strip(),
        "retrieved_data": records,
        "question": f"پرسش: {question_text}\nگزینه‌ها:\n" + "\n".join(options),
    }
```

```python
def rag_qa_from_text(question_text: str, query: str, options, k_text=5, k_image=5):
    t_hits, i_hits = retrieve_for_text_query(query, k_text=k_text, k_image=k_image)
    return gemma_rag_generate(question_text, options, t_hits, i_hits)

def rag_qa_from_image(question_text: str, query: str, options, k_text=5, k_image=5):
    t_hits, i_hits, url = retrieve_for_image_query(query, k_text=k_text, k_image=k_image)
    return gemma_rag_generate(question_text, options, t_hits, i_hits, url)
```

```python
def find_best_option(answer, options):
    """Match model's answer to the closest option."""
    match = get_close_matches(answer, options, n=1, cutoff=0.3)
    return match[0] if match else None

def run_pipeline(input_path, output_path):
    with open(input_path, "r", encoding="utf-8") as f:
        image_type2_questions = json.load(f)
    results = []
    for q in tqdm(image_type2_questions):
        question_text = q["question"]
        options = q["options"]
        correct_answer = q['answer']

        if q['image'] is not "null":
            query = q['image']
            result = rag_qa_from_image(question_text, query, options, k_text=0, k_image=5)
        else:
            print(question_text+ " " + ", ".join(options))
            query = question_text+ " " + ", ".join(options)
            result = rag_qa_from_text(question_text, query, options, k_text=5, k_image=0)
            
        predicted_option = find_best_option(result["answer"], options)

        result = {
            "id": q["id"],
            "question": question_text+ " " + ", ".join(options),
            "retrieved_data": result["retrieved_data"],
            "raw_answer": result["answer"],
            "selected_answer": predicted_option,
            "correct_answer": correct_answer
        }

        print(json.dumps(result, indent=4, ensure_ascii=False))
        print('-'*100)
        
        with open(output_path, "a", encoding="utf-8") as f:
            json.dump(result, f, ensure_ascii=False, indent=2)
```

```python
from google import genai
import time

api_key1 = ""
api_key2 = ""
API_KEYS = [api_key1, api_key2]
current_key_index = 0

client = genai.Client(api_key = API_KEYS[current_key_index])

def ask_gemini(user_input):
    global client, current_key_index
    while True:
        try:
            response = client.models.generate_content(
                model="gemini-2.0-flash",  
                contents=user_input
            )
            return (response.text)
        except:
            current_key_index = (current_key_index + 1) % len(API_KEYS)
            new_key = API_KEYS[current_key_index]
            print(f"🔑 Swapping to API key index {current_key_index}")

            client = genai.Client(api_key=new_key)
```

```python
import json
from tqdm import tqdm


def evaluate_on(llm_model, input_path, output_path):
    with open(input_path, "r", encoding="utf-8") as f:
        questions = json.load(f)

    for q in tqdm(questions):
        question_text = "فقط یکی از گزینه های زیر را انتخاب کن. "+q["question"]
        options = "\n".join(q["options"])
        correct_answer = q['answer']
        query = question_text+"\n"+options
        result = llm_model(query)
            
        # predicted_option = find_best_option(result, q["options"])

        final_result = {
            "id": q["id"],
            "question": question_text,
            # "retrieved_data": result["retrieved_data"],
            "raw_answer": result,
            # "selected_answer": predicted_option,
            "correct_answer": correct_answer
        }

        print(f"\n{query} ")
        print('\nresult: ', result)
        print('\ncorrect answer: ', correct_answer)
        print('-'*60)
        
        with open(output_path, "a", encoding="utf-8") as f:
            json.dump(final_result, f, ensure_ascii=False, indent=2)
```
